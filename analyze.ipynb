```python
import pandas as pd
import matplotlib.pyplot as plt

path = "/home/ktang/ashi_data/21918746/HD2_ProP-PD_unfiltered_peptides/HD2_ProP-PD_unfiltered_peptides.tsv"

df = pd.read_csv(path, sep="\t", usecols=["pls_supervised", "pls_supervised_text"])

# --- Unique entries in pls_supervised_text ---
counts = df["pls_supervised_text"].value_counts()
print(counts.to_string())
print("Unique entries:", df["pls_supervised_text"].nunique())

# --- Histogram of pls_supervised (N/A excluded) ---
vals = pd.to_numeric(df["pls_supervised"], errors="coerce")
n_na = vals.isna().sum()
vals = vals.dropna()
print(f"\nExcluded N/A: {n_na} | Plotted: {len(vals)}")
print(vals.describe())

plt.figure(figsize=(8, 5))
plt.hist(vals, bins=100, color="steelblue", edgecolor="black", linewidth=0.3)
plt.xlabel("pls_supervised")
plt.ylabel("Count")
plt.title("Histogram of pls_supervised")
plt.tight_layout()
plt.savefig("/home/ktang/ashi_data/pls_supervised_histogram.png", dpi=150)
plt.show()
```

```python
import pandas as pd

path = "/home/ktang/ashi_data/21918746/HD2_ProP-PD_unfiltered_peptides/HD2_ProP-PD_unfiltered_peptides.tsv"

df = pd.read_csv(path, sep="\t", usecols=["hit_accession", "pls_supervised_text"])

order = ["N/A", "low", "medium", "high", "excellent"]
rank = {t: i for i, t in enumerate(order)}

# Keep the best label per unique hit_accession
best = (
    df.assign(_rank=df["pls_supervised_text"].map(rank))
      .sort_values("_rank")
      .groupby("hit_accession", as_index=False)
      .last()[["hit_accession", "pls_supervised_text"]]
)

counts = best["pls_supervised_text"].value_counts().reindex(order).dropna().astype(int)
print("Unique hit_accession:", best["hit_accession"].nunique())
print(counts.to_string())
```
